## Qlib表达式函数完整列表

### 1. **数学运算函数**
- `Add($field1, $field2)` - 加法运算
- `Sub($field1, $field2)` - 减法运算  
- `Mul($field1, $field2)` - 乘法运算
- `Div($field1, $field2)` - 除法运算
- `Pow($field, n)` - 幂运算
- `Sqrt($field)` - 平方根
- `Abs($field)` - 绝对值
- `Log($field)` - 对数
- `Exp($field)` - 指数

### 2. **统计函数**
- `Max($field, n)` - n期最大值
- `Min($field, n)` - n期最小值
- `Sum($field, n)` - n期求和
- `Std($field, n)` - n期标准差
- `Var($field, n)` - n期方差
- `Skew($field, n)` - n期偏度
- `Kurt($field, n)` - n期峰度

### 3. **时间序列函数**
- `Ref($field, n)` - 向前引用n期
- `Delta($field, n)` - n期变化量
- `Rank($field, n)` - n期排名
- `Quantile($field, n, q)` - n期分位数
- `Count($field, n)` - n期非空值计数

### 4. **相关性函数**
- `Corr($field1, $field2, n)` - n期相关系数
- `Cov($field1, $field2, n)` - n期协方差

### 5. **技术指标函数**
- `MA($field, n)` - n期移动平均（与Mean相同）
- `EMA($field, n)` - n期指数移动平均
- `MACD($field, fast, slow, signal)` - MACD指标
- `RSI($field, n)` - RSI相对强弱指标
- `Bollinger($field, n, k)` - 布林带指标

### 6. **条件函数**
- `If($condition, $true_value, $false_value)` - 条件判断
- `Where($condition, $field)` - 条件筛选

### 7. **基础字段**
- `$open` - 开盘价
- `$close` - 收盘价
- `$high` - 最高价
- `$low` - 最低价
- `$volume` - 成交量
- `$amount` - 成交额
- `$pb` - 市净率
- `$pe` - 市盈率




## D.features() 函数参数

`D.features()` 函数的主要参数包括：
- `instruments`: 股票代码列表
- `fields`: 字段列表（如 `['$close', '$volume']`）
- `start_time`: 开始时间
- `end_time`: 结束时间  
- `freq`: 频率（如 'day', '1min'）

## features对象（DataFrame）的常用方法

由于`D.features()`返回的是一个pandas DataFrame，它具有所有DataFrame的常用方法：

### 1. 基本信息查看
```python
features.shape          # 数据形状 (行数, 列数)
features.head()         # 查看前5行
features.tail()         # 查看后5行
features.info()         # 数据信息
features.describe()     # 统计描述
features.columns        # 列名
features.index          # 索引信息
```

### 2. 数据选择和过滤
```python
features['$close']      # 选择特定列
features[['$close', '$volume']]  # 选择多列
features.loc[date]      # 按日期选择
features.iloc[0:10]     # 按位置选择
features.dropna()       # 删除空值
features.fillna(0)      # 填充空值
```

### 3. 数据操作
```python
features.rename(columns={'$close': 'close'})  # 重命名列
features.reset_index()  # 重置索引
features.set_index('date')  # 设置索引
features.sort_values('$close')  # 排序
```

### 4. 统计计算
```python
features.mean()         # 均值
features.std()          # 标准差
features.min()          # 最小值
features.max()          # 最大值
features.sum()          # 求和
features.count()        # 计数
features.quantile(0.5)  # 分位数
```

### 5. 分组操作
```python
features.groupby(level=1).mean()  # 按股票分组计算均值
features.groupby(level=0).sum()   # 按日期分组计算总和
```

### 6. 数据转换
```python
features.pct_change()   # 计算变化率
features.diff()         # 计算差分
features.shift(1)       # 数据平移
features.rolling(5).mean()  # 滚动窗口计算
```

### 7. 数据质量检查
```python
features.isnull().sum()  # 检查空值数量
features.duplicated().sum()  # 检查重复值
features.nunique()       # 唯一值数量
```

### 8. 保存和导出
```python
features.to_csv('data.csv')     # 保存为CSV
features.to_excel('data.xlsx')  # 保存为Excel
features.to_pickle('data.pkl')  # 保存为pickle
```

# 回测结果分析

1.  回测频率键：portfolio_metric_dict 的键是 ['1day']，这表示你的回测是按日度（'1day'）频率进行的。你正确地使用 '1day' 作为键来提取了 report_df (绩效报告DataFrame) 和 positions_dict (持仓字典)。

2.  绩效报告 (report_df)：
    *   report_df.head() 显示的是回测开始后前几个交易日的账户概况。
    *   各列含义：
        *   account: 当日收盘后的账户总价值（单位：元）。初始值为 100,000,000（1亿），随后每日波动。
        *   return: 投资组合的日收益率。
        *   total_turnover: 累计总交易额（自回测开始以来的总和）。
        *   turnover: 当日交易额占账户价值的比例（即换手率）。
        *   total_cost: 累计总交易成本（手续费等）。
        *   cost: 当日交易成本。
        *   value: 持仓市值。
        *   cash: 现金余额。
        *   bench: 基准指数（沪深300）的日收益率。

3.  持仓详情 (positions_dict)：
    *   这是一个以日期为键，以该日期对应的持仓对象为值的字典。
    *   你提取的样本日期是 '2015-01-06 00:00:00'，其持仓详情的类型是 <class 'qlib.backtest.position.Position'>，这是一个 Qlib 内部表示持仓的类对象。
    *   打印这个对象显示，它内部包含了一个字典，其中除了结算类型 (_settle_type) 外，核心是 'position' 字典。
    *   'position' 字典里包含了：
        *   cash: 现金余额。
        *   now_account_value: 当前账户总价值。
        *   各个股票代码（如 '600637.SH'）：每个股票代码下又是一个字典，包含：
            *   amount: 持仓数量（股数）。
            *   price: 当前价格（元）。
            *   weight: 该股票在投资组合中的权重。
            *   count_day: 该持仓已持有的天数。
