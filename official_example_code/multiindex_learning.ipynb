# 1.创建多级索引

```python
import pandas as pd
import numpy as np

# 从元组列表创建
arrays = [['A', 'A', 'B', 'B'], [1, 2, 1, 2]]
tuples = list(zip(*arrays))
index = pd.MultiIndex.from_tuples(tuples, names=['letters', 'numbers'])
df_multi = pd.DataFrame(np.random.randn(4, 2), index=index, columns=['col1', 'col2'])
print(df_multi)
```

```python
# 从数组创建
arrays = [np.array(['bar', 'bar', 'baz', 'baz']), np.array(['one', 'two', 'one', 'two'])]
df_multi = pd.DataFrame(np.random.randn(4, 2), index=arrays)
print(df_multi)
```

```python
# 从DataFrame创建
df_multi = pd.DataFrame({
    'A': ['foo', 'foo', 'bar', 'bar'],
    'B': ['one', 'two', 'one', 'two'],
    'C': [1, 2, 3, 4],
    'D': np.random.randn(4)
})

df_multi = df_multi.set_index(['A', 'B'])
print(df_multi)
```

# 2. 索引基本操作

```python
import pandas as pd
import numpy as np

df_multi=pd.DataFrame({
    'letters':['A','B','C','D'],
    'numbers':['1','2','3','4'],
    'value': np.random.randn(4)
})
print(df_multi)
df_multi=df_multi.set_index(['letters','numbers'])
print(df_multi)
```

```python
print(type(df_multi.index))  
# <class 'pandas.core.indexes.multi.MultiIndex'>

print(df_multi.index.names)  
# ['level1', 'level2']

print(df_multi.index.levels)  
# FrozenList([['A', 'B'], [1, 2]])
```

```python
# 查看索引属性
print("\n=== 索引属性 ===")
print("索引类型:", type(df_multi.index))
print("索引名称:", df_multi.index.names)
print("索引层级:", df_multi.index.levels)

# 重置索引
print("\n=== 重置索引 ===")
df_reset = df_multi.reset_index()
print(df_reset)

# 交换层级
print("\n=== 交换层级 ===")
df_swapped = df_multi.swaplevel('letters', 'numbers')
print(df_swapped)

# 排序索引
print("\n=== 排序后 ===")
df_sorted = df_multi.sort_index()
print(df_sorted)
```

# 3.数据选择与切片

```python
import pandas as pd
import numpy as np

df_multi=pd.DataFrame({
    'letters':['A','B','C','D'],
    'numbers':[1,2,3,4],
    'value': np.random.randn(4),
    'attribute':['big','small','big','small']
})
print(df_multi)
df_multi=df_multi.set_index(['letters','numbers'])
print(df_multi)
```

```python
# 使用xs选择特定层级数据
df_multi.xs('A', level='letters')  # 选择letters层级为'A'的所有数据
df_multi.xs(('A', 1), level=['letters', 'numbers'])  # 选择特定组合
```

```python
# 使用loc进行选择
df_multi.loc['A']  # 选择第一层级为'A'的所有数据
df_multi.loc[('A', 1)]  # 选择特定组合
df_multi.loc[('A', 1):('B', 2)]  # 切片选择
```

```python
# 使用pd.IndexSlice进行复杂切片
idx = pd.IndexSlice
df_multi.loc[idx[:, 1], :]  # 选择所有第一层级，第二层级为1的数据
df_multi.loc[idx['A', :], :]  # 选择第一层级为'A'的所有数据
df_multi.loc[idx['A':'B', 1:2], :]  # 范围切片
```

```python
# 选择特定列

print(df_multi.loc[:, 'value'] ) # 选择单列

print(df_multi.loc[:, ['value', 'attribute']] ) # 选择多列
```

# 4. 数据查询与过滤

```python
import pandas as pd
import numpy as np

df_multi=pd.DataFrame({
    'letters':['A','B','C','D'],
    'numbers':[1,2,3,4],
    'value': np.random.randn(4),
    'attribute':['big','small','big','small']
})
print(df_multi)
df_multi=df_multi.set_index(['letters','numbers'])
print(df_multi)
print(df_multi.index.names)
```

```python
# 按条件过滤
df_multi[df_multi['value'] > 0]  # 普通条件过滤

# 多层级条件查询
mask = (df_multi.index.get_level_values('letters') == 'A') & (df_multi.index.get_level_values('numbers') == 1)
df_multi[mask]

# 使用query方法
df_multi.query('letters == "A" and numbers == 1')
```

# 5.数据重塑与转换

```python
# 堆叠与解堆叠
stacked = df_multi.stack()  # 将列转换为行索引
unstacked = df_multi.unstack()  # 将行索引转换为列
unstacked_level = df_multi.unstack('letters')  # 指定要解堆叠的层级

# 数据透视表
pivoted = df_multi.pivot_table(values='D', index='A', columns='B', aggfunc='mean')

# 数据融合
melted = pd.melt(df_multi, id_vars=['A', 'B'], value_vars=['C', 'D'])
```

# 6.数据分组与聚合

```python
# 按层级分组聚合
df_multi.groupby(level='letters').mean()
df_multi.groupby(level=[0, 1]).sum()  # 按索引位置分组

# 多层级分组
df_multi.groupby(['A', 'B']).mean()

# 使用agg进行多种聚合
df_multi.groupby(level='letters').agg({'col1': 'mean', 'col2': 'sum'})

# 使用transform进行分组转换
df_multi['group_mean'] = df_multi.groupby(level='letters')['col1'].transform('mean')
```

# 7.索引操作与修改

```python
# 重命名索引层级
df_multi.index = df_multi.index.set_names(['new_name1', 'new_name2'])

# 添加新层级
df_new = df_multi.set_index('new_level', append=True)

# 删除层级
df_dropped = df_multi.droplevel('letters')

# 重组索引
new_index = pd.MultiIndex.from_product([['A', 'B'], [1, 2]], names=['letters', 'numbers'])
df_reindexed = df_multi.reindex(new_index)

# 填充缺失值
df_filled = df_reindexed.fillna(0)
```

# 8.多级索引计算

```python
# 跨层级计算
df_multi['new_col'] = df_multi['col1'] + df_multi['col2']

# 分组计算百分比
df_multi['pct'] = df_multi.groupby(level='letters')['col1'].apply(lambda x: x / x.sum())

# 计算层级间差值
df_multi['diff'] = df_multi.groupby(level='letters')['col1'].diff()

# 计算移动平均
df_multi['ma'] = df_multi.groupby(level='letters')['col1'].rolling(window=3).mean().values
```
