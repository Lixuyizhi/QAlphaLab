# 公司市值与高管平均薪酬的相关关系

```python
import pandas as pd
import numpy as np
```

```python
df_marketvalue=pd.read_csv('./.qlib/mid_data/ashare_marketvalue.csv')
df_mean_holdreward=pd.read_csv('./.qlib/mid_data/ashare_mean_holdreward.csv')
df_aindexmember= pd.read_csv('./.qlib/mid_data/AINDEXMEMBERS.csv')
```

```python

df_mean_holdreward.rename(columns={'ANN_DATE':'datetime'},inplace=True)
df_mean_holdreward.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df_marketvalue.info()
df_mean_holdreward.info()
```

```python
df_combined=df_mean_holdreward.merge(df_marketvalue, on=['code','datetime'], how='inner')
```

```python

df_combined = df_combined.dropna(subset=['marketvalue'])
```

```python
df_combined
```

```python
allashare_spearman_corr = df_combined['mean_return'].corr(df_combined['marketvalue'], method='spearman')
```

```python
allashare_spearman_corr
```

## 细分指数

```python
df_aindexmember
```

```python
df_combined=df_combined.merge(df_aindexmember[['code','S_IRDCODE']], on=['code'], how='inner')
```

```python
# 上证50
shangzheng50_combined=df_combined[df_combined['S_IRDCODE']=='000016.SH']
shangzheng50_spearman_corr = shangzheng50_combined['mean_return'].corr(shangzheng50_combined['marketvalue'], method='spearman')
print(f'上证50中高管薪酬和公司市值的相关系数{shangzheng50_spearman_corr}\n')
```

```python
# 沪深300
hushen300_combined=df_combined[df_combined['S_IRDCODE']=='000300.SH']
hushen300_combined_spearman_corr = hushen300_combined['mean_return'].corr(hushen300_combined['marketvalue'], method='spearman')
print(f'沪深300中高管薪酬和公司市值的相关系数{hushen300_combined_spearman_corr}\n')
```

```python
# 中证500
zhongzheng500_combined=df_combined[df_combined['S_IRDCODE']=='000905.SH']
zhongzheng500_combined_spearman_corr = zhongzheng500_combined['mean_return'].corr(zhongzheng500_combined['marketvalue'], method='spearman')
print(f'中证500中高管薪酬和公司市值的相关系数{zhongzheng500_combined_spearman_corr}\n')
```
